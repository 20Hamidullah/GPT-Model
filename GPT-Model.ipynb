## 2. Perplexity

Given: the sentence `"The students open their books unwillingly"`, and the following word probabilities from two models:

| Word        | Model A | Model B |
|-------------|---------|---------|
| The         | 0.25    | 0.10    |
| students    | 0.30    | 0.05    |
| open        | 0.20    | 0.10    |
| their       | 0.10    | 0.10    |
| books       | 0.10    | 0.05    |
| unwillingly | 0.05    | 0.01    |

#### Tasks:

- Calculate the perplexity for each model using:

$$
\text{PPL} = \exp\left( -\frac{1}{N} \sum_{i=1}^{N} \log P(w_i|w_1,...,w_{i-1}) \right)
$$



- Which model performs better? Why?

Calculation of Perplexity for each of the model using the given formula:

−log(0.25) = 1.386, −log(0.25)=1.386, −log(0.30)= 1.204, −log(0.30)=1.204, −log(0.20)=1.609, −log(0.20)=1.609, −log(0.10)=2.302, −log(0.10)=2.302, −log(0.10)=2.302, −log(0.10)=2.302, −log(0.05)= 2.996, −log(0.05)=2.996.

Average: (1.386+1.204+1.609+2.302+2.302+2.996)/6 ≈ 1.967

Perplexity: exp(1.967) ≈ 7.15

Model B:

⁡(0.10)=2.302, −log(0.10)=2.302, −log(0.05)=2.996, −log(0.05)=2.996, −log(0.10)=2.302, −log(0.10)=2.302, −log(0.10)=2.302, −log(0.10)=2.302, −log(0.05)=2.996, −log(0.05)=2.996, −log(0.01)=4.605, −log(0.01)=4.605.

Conclusion:

## 3. GPT model
we will use the GPT model for text generation. We start with the **transformer** library. This library allows us to access multiple pre-trained models for you to choose from.

```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel
import torch

# Text Generation with GPT-2 as Decoder
def generate_text(prompt):
    # Initialize tokenizer and models
    tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
    model = GPT2LMHeadModel.from_pretrained("gpt2")
    model.eval()

    ## convert a text prompt into token IDs that can be used as input for a GPT-2 model, return a Pytorch tensor
    gpt2_inputs = tokenizer.encode(prompt, return_tensors="pt", max_length=1024, truncation=True)
    #Create an attention mask, and filled with 1-real(0-padding)
    attention_mask = torch.ones_like(gpt2_inputs)
    gpt2_outputs = model.generate(input_ids=gpt2_inputs,
                                       max_length=1024,
                                       no_repeat_ngram_size=2, #Prevents the model from repeating 2-grams (or any n-gram of that size) in the generated text.
                                       num_beams=5, #Enables beam search decoding (keeps track of the top k most likely sequences) instead of greedy decoding (onely one best value)
                                       early_stopping=True, #Stop when the best sequence is complete
                                       attention_mask=attention_mask, #Tells the model which tokens in the input are actual content (1) and which are padding (0).
                                       pad_token_id=tokenizer.eos_token_id) # GPT-2 does not have a pad token by default, so here we explicitly set the pad token to be the same as the end-of-sequence token

    # Decode the output from GPT-2, convert output to the string, removes special tokens
    generated_text = tokenizer.decode(gpt2_outputs[0], skip_special_tokens=True)

    return generated_text

# Let users enter the prompt
prompt = input("Enter your prompt: ")
generated_text = generate_text(prompt)
print("Generated Text:", generated_text)
```

### Tasks:
1. Replace beam search with top-k or top-p sampling and compare results.

```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel
import torch

def generate_text(prompt, method='beam'):
    # Initialize tokenizer and models
    tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
    model = GPT2LMHeadModel.from_pretrained("gpt2")
    model.eval()

    gpt2_inputs = tokenizer.encode(prompt, return_tensors="pt", max_length=1024, truncation=True)
    attention_mask = torch.ones_like(gpt2_inputs)

    if method == 'beam':
        outputs = model.generate(
            input_ids=gpt2_inputs,
            max_length=1024,
            num_beams=5,
            no_repeat_ngram_size=2,
            early_stopping=True,
            attention_mask=attention_mask,
            pad_token_id=tokenizer.eos_token_id
        )
    elif method == 'topk':
        outputs = model.generate(
            input_ids=gpt2_inputs,
            do_sample=True,
            top_k=50,
            temperature=0.7,
            max_length=1024,
            attention_mask=attention_mask,
            pad_token_id=tokenizer.eos_token_id
        )
    elif method == 'topp':
        outputs = model.generate(
            input_ids=gpt2_inputs,
            do_sample=True,
            top_p=0.92,
            temperature=0.7,
            max_length=1024,
            attention_mask=attention_mask,
            pad_token_id=tokenizer.eos_token_id
        )

    return tokenizer.decode(outputs[0], skip_special_tokens=True)

# Compare methods
prompt = input("Enter your prompt: ")

print("\nBeam Search:")
print(generate_text(prompt, method='beam'))

print("\nTop-k Sampling:")
print(generate_text(prompt, method='topk'))

print("\nTop-p Sampling:")
print(generate_text(prompt, method='topp'))
```

```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel
import torch
import torch.nn.functional as F

def show_top_predictions(prompt, top_k=5):
    tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
    model = GPT2LMHeadModel.from_pretrained("gpt2")
    model.eval()

    # Tokenize input
    inputs = tokenizer(prompt, return_tensors="pt")

    # Forward pass
    with torch.no_grad():
        outputs = model(**inputs)

    # Get logits for last token
    logits = outputs.logits[:, -1, :]
    probs = F.softmax(logits, dim=-1).squeeze()

    # Get top predictions
    top_probs, top_indices = torch.topk(probs, top_k)

    print(f"\nTop {top_k} predictions for: '{prompt}'")
    for i, (prob, idx) in enumerate(zip(top_probs, top_indices)):
        word = tokenizer.decode(idx)
        print(f"{i+1}. {word!r}: {prob:.2%}")

# Example usage
show_top_predictions("The future of AI will")
```
